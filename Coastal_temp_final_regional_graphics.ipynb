## Overview
NOAA's CO-OPS collects trusted coastal data—including water temperature, tides, and currents—that helps keep navigation safe, supports commerce, and protects coastal communities and ecosystems. Since the early 1990s, over 220 water temperature sensors have been installed along U.S. coasts and in the Great Lakes. CO-OPS did an in-depth analysis on 174 stations with at least 20 years of data. An automated quality control system was developed to streamline data processing, flagging suspect data, and calculating daily averages. Manual reviews also caught sensor drifts at 47 stations, ensuring the data is reliable for long-term research.

  ## Data Sources and Documentation

  All data comes from CO-OPS's sensors. Stations with water temperature data can be found via [CO-OPS Metadata API](https://api.tidesandcurrents.noaa.gov/mdapi/prod/webapi/stations.json?type=watertemp). 6-min water temperature data can be accessed through [CO-OPS Data API](https://api.tidesandcurrents.noaa.gov/api/prod/).
  
  The QA/QCed daily water temperature data is currently available on github with this notebook. Each station’s data is provided in a separate comma-separated values (CSV) file, containing daily means, residuals, and quality control (QC) flags. Each CSV includes the following columns: Date, DAILY_WATER_TEMP, residual, manual_flag, automated_flag, and data_fill_flag. Suspect data in the manual and automated flag columns are marked with a “1,” while data that passed QC remain unflagged. The data_fill_flag column indicates the source of any replacement data (e.g., DCP2 E1, DCP3 E1, DCP1 E2) on the dates it was used, while unfilled data have no entry in this column.

## **Step 4**

- Read the csv file
- Remove flagged data
- Plot and save graphics

```python
# @title Code to product cleaned regional comparison plots
if region_choice == 'single':
    station_list = [region_selection]
for region in station_list:
    print(str(region))
    station_region = station_region_list[station_region_list['Group Median Regions']==region]
    print(station_region)
    temp_dfs = {}  # Create an empty dictionary to store the dataframes
    resid_dfs = {}
    for i in range(len(station_region)):
        station = station_region['Station ID'].iloc[i]
        try:
            station_df = pd.read_csv(path+str(station)+'_final_daily_flagged.csv')
        except FileNotFoundError:
            print('File not found')
            continue
        #removing flagged data
        try:
            station_df.loc[station_df['manual_flag'] == 1, 'DAILY_WATER_TEMP'] = np.nan
            station_df.loc[station_df['manual_flag'] == 1, 'residual'] = np.nan
        except KeyError:
                print('No manual flag for this station')
        try:
            station_df.loc[station_df['automated_flag'] == 1, 'DAILY_WATER_TEMP'] = np.nan
            station_df.loc[station_df['automated_flag'] == 1, 'residual'] = np.nan
        except KeyError:
            print('No automated flag for this station')
        temp_df = station_df[['Date','DAILY_WATER_TEMP']]
        resid_df = station_df[['Date','residual']]
        temp_df=temp_df.rename(columns={'DAILY_WATER_TEMP':station})
        resid_df = station_df[['Date','residual']]
        temp_df=temp_df.rename(columns={'DAILY_AIR_TEMP':station})
        temp_df.loc[:, 'Date'] = pd.to_datetime(temp_df['Date'])
        resid_df.loc[:, 'Date'] = pd.to_datetime(resid_df['Date'])
        resid_df=resid_df.rename(columns={'residual':station})
        temp_df=temp_df.drop_duplicates()
        resid_df=resid_df.drop_duplicates()
        temp_df=temp_df.sort_values(by='Date')
        resid_df=resid_df.sort_values(by='Date')
        key = f"station_{i+1}"  # Create a key for the dictionary, e.g., "station_1", "station_2", ...
        temp_dfs[key] = temp_df  # Add the dataframe to the dictionary with the key
        resid_dfs[key] = resid_df
    # List of keys for the station dataframes in station_dfs dictionary
    keys = [f"station_{i+1}" for i in range(len(station_region))]

    # List of dataframes from the station_dfs dictionary
    data_frames_temp = [temp_dfs[key] for key in keys]
    data_frames_resid = [resid_dfs[key] for key in keys]

    # Merge all dataframes in the list on 'Date' column
    combined_df_temp = reduce(lambda left, right: pd.merge(left, right, on=['Date'], how='outer'), data_frames_temp)
    combined_df_temp = combined_df_temp.sort_values(by='Date')
    combined_df_temp = combined_df_temp.set_index(combined_df_temp['Date'],drop=True)

    combined_df_resid = reduce(lambda left, right: pd.merge(left, right, on=['Date'], how='outer'), data_frames_resid)
    combined_df_resid = combined_df_resid.sort_values(by='Date')
    combined_df_resid = combined_df_resid.set_index(combined_df_resid['Date'],drop=True)

    numsta = len(station_region)
    station_columns = combined_df_temp.columns[1:numsta+1]  # Select station columns for plotting

    # Create a list of traces (one for each station)
    traces = []
    for station in station_columns:
        trace = go.Scatter(
            x=combined_df_temp['Date'],
            y=combined_df_temp[station],
            mode='lines',
            name=str(station)  # Use station ID as the name
        )
        traces.append(trace)

    # Create the figure with the traces
    fig = go.Figure(data=traces)

    # Update layout
    fig.update_layout(
        width=1400,  # width of the plot in pixels
        height=400,
        title=station_region['Region Name'].iloc[0] + ' Stations Daily Temperature',
        title_x=0.5,  # Center the title (x=0.5 is the center)
        title_y=0.94,  # Optional: adjust the vertical position of the title
        title_font=dict(size=25),  # Increase font size of the title
        yaxis_title='Temperature (°C)',  # Y-axis label
        xaxis_title='Years',  # Optional: Add X-axis label if needed
        plot_bgcolor='white',  # Set plot background to white
        xaxis=dict(
            gridcolor='black',  # Set X-axis gridline color to black
            #showgrid=True,  # Show X-axis gridlines
            # tickangle=45  # Rotate X-axis ticks if needed
            tickfont=dict(size=25),
            title_font=dict(size=25)
        ),
        yaxis=dict(
            gridcolor='black',  # Set Y-axis gridline color to black
          # showgrid=True  # Show Y-axis gridlines
            tickfont=dict(size=25),
            title_font=dict(size=25)
        ),
        legend_title_text='Stations',  # Legend title text
        legend=dict(
            title_font=dict(size=18),  # Set font size for the legend title
            font=dict(size=16),  # Set font size for the legend items
        )
    )
    #save interactive plot as html file
    fig_path = r'Regional_Final_Plots/'+str(station_region['Group Median Regions'].iloc[0])
    if not os.path.exists(fig_path):
        os.makedirs(fig_path)
    fig.write_html('Regional_Final_Plots/'+str(station_region['Group Median Regions'].iloc[0])+'/'+str(station_region['Group Median Regions'].iloc[0])+'_daily_timeseries.html')
    fig.show()

    combined_df_temp['Median'] = combined_df_temp.iloc[:, 1:numsta+1].median(axis=1)

    fig2 = px.line(combined_df_temp, x='Date', y=combined_df_temp['Median'])

    # Update the layout
    fig2.update_layout(
        width=1400,  # width of the plot in pixels
        height=400,
        title=station_region['Region Name'].iloc[0] + ' Regional Median Temperature',
        title_x=0.5,  # Center the title (x=0.5 is the center)
        title_y=0.94,  # Optional: adjust the vertical position of the title
        title_font=dict(size=25),  # Increase font size of the title
        yaxis_title='Temperature (°C)',  # Y-axis label
        xaxis_title='Years',  # Optional: Add X-axis label if needed
        plot_bgcolor='white',  # Set plot background to white
        xaxis=dict(
            gridcolor='black',  # Set X-axis gridline color to black
            #showgrid=True,  # Show X-axis gridlines
            # tickangle=45  # Rotate X-axis ticks if needed
            tickfont=dict(size=25),
            title_font=dict(size=25)
        ),
        yaxis=dict(
            gridcolor='black',  # Set Y-axis gridline color to black
          # showgrid=True  # Show Y-axis gridlines
            tickfont=dict(size=25),
            title_font=dict(size=25)
        ),
        legend_title_text='Stations',  # Legend title text
        legend=dict(
            title_font=dict(size=18),  # Set font size for the legend title
            font=dict(size=16),  # Set font size for the legend items
        )
    )

    fig2.write_html('Regional_Final_Plots/'+str(station_region['Group Median Regions'].iloc[0])+'/'+str(station_region['Group Median Regions'].iloc[0])+'_daily_timeseries.html')
    fig2.show()

    for i in list(range(1,numsta+1)):
      exclude_median= pd.DataFrame(combined_df_resid.iloc[:,0])
      for j in list(range(1,numsta+1)):
        if not j == i :
          exclude_median[combined_df_resid.columns[j]] = combined_df_resid.iloc[:,j]
        combined_df_resid['group median w/o ' + str(combined_df_resid.columns[i])] = exclude_median.median(axis=1,numeric_only=True)
    for i in list(range(1,numsta+1)):
      combined_df_resid[str(combined_df_resid.columns[i]) + ' - group median'] = combined_df_resid.iloc[:,i] - combined_df_resid.iloc[:,i+numsta]
    combined_df_resid['median of group medians'] = combined_df_resid.iloc[:,numsta+1:2*numsta+1].median(axis=1,numeric_only=True)

    fig3 = px.line(combined_df_resid, x='Date', y=combined_df_resid.columns[[*range(numsta+1,2*numsta+1)]])
    fig3.update_traces(opacity=0.8)
    fig3.update_yaxes(zerolinecolor='black', zerolinewidth=1.5)

    fig3.add_scatter(x=combined_df_resid['Date'], y=combined_df_resid.iloc[:,3*numsta+1], mode='lines', line_color='black', name = 'median of group medians')

    # Update layout
    # Update the layout
    fig3.update_layout(
        width=1400,  # width of the plot in pixels
        height=400,
        title=station_region['Region Name'].iloc[0] + ' Group Median Residuals where single station is excluded',
        title_x=0.5,  # Center the title (x=0.5 is the center)
        title_y=0.94,  # Optional: adjust the vertical position of the title
        title_font=dict(size=25),  # Increase font size of the title
        yaxis_title='Temperature (°C)',  # Y-axis label
        xaxis_title='Years',  # Optional: Add X-axis label if needed
        plot_bgcolor='white',  # Set plot background to white
        xaxis=dict(
            gridcolor='black',  # Set X-axis gridline color to black
            #showgrid=True,  # Show X-axis gridlines
            # tickangle=45  # Rotate X-axis ticks if needed
            tickfont=dict(size=25),
            title_font=dict(size=25)
        ),
        yaxis=dict(
            gridcolor='black',  # Set Y-axis gridline color to black
          # showgrid=True  # Show Y-axis gridlines
            tickfont=dict(size=25),
            title_font=dict(size=25)
        ),
        legend_title_text='Stations',  # Legend title text
        legend=dict(
            title_font=dict(size=18),  # Set font size for the legend title
            font=dict(size=16),  # Set font size for the legend items
        )
    )
    fig3.update_layout(legend_title_text='Group Median')
    fig3.write_html('Regional_Final_Plots/'+str(station_region['Group Median Regions'].iloc[0])+'/'+str(station_region['Group Median Regions'].iloc[0])+'_regional_residual_median_plot.html')
    fig3.show()

    fig4 = px.line(combined_df_resid, x='Date', y=combined_df_resid.columns[2*numsta+1 : 3*numsta+1])

    fig4.update_layout(
        width=1400,  # width of the plot in pixels
        height=400,
        title=dict(text=station_region['Region Name'].iloc[0] + ' Station Difference from Group Median', font=dict(size=25)),

        title_x=0.5,  # Center the title (x=0.5 is the center)
        title_y=0.94,  # Optional: adjust the vertical position of the title
        title_font=dict(size=25),  # Increase font size of the title
        yaxis_title='Temperature (°C)',  # Y-axis label
      # xaxis_title='Years',  # Optional: Add X-axis label if needed
        plot_bgcolor='white',  # Set plot background to white
        xaxis=dict(
        gridcolor='black',  # Set X-axis gridline color to black
                        #showgrid=True,  # Show X-axis gridlines
                        # tickangle=45  # Rotate X-axis ticks if needed
        tickfont=dict(size=20),
        title_font=dict(size=20)
        ),
        yaxis=dict(
        gridcolor='black',  # Set Y-axis gridline color to black
                    # showgrid=True  # Show Y-axis gridlines
        tickfont=dict(size=20),
        title_font=dict(size=20)
        ),
        legend_title_text='Station Difference from Group Median',  # Legend title text
        legend=dict(
        title_font=dict(size=18),  # Set font size for the legend title
        font=dict(size=16),  # Set font size for the legend items
    ))
    fig4.update_layout(legend_title_text='Group Median')
    fig4.write_html('Regional_Final_Plots/'+str(station_region['Group Median Regions'].iloc[0])+'/'+str(station_region['Group Median Regions'].iloc[0])+'_regional_residual_difference_plot.html')
    fig4.show()
```
