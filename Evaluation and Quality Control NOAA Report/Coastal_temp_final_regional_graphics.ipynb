## Overview
NOAA's CO-OPS collects trusted coastal data—including water temperature, tides, and currents—that helps keep navigation safe, supports commerce, and protects coastal communities and ecosystems. Since the early 1990s, over 220 water temperature sensors have been installed along U.S. coasts and in the Great Lakes. CO-OPS did an in-depth analysis on 174 stations with at least 20 years of data. An automated quality control system was developed to streamline data processing, flagging suspect data, and calculating daily averages. Manual reviews also caught sensor drifts at 47 stations, ensuring the data is reliable for long-term research.

## Data Sources and Documentation

All data comes from CO-OPS's sensors. Stations with water temperature data can be found via [CO-OPS Metadata API](https://api.tidesandcurrents.noaa.gov/mdapi/prod/webapi/stations.json?type=watertemp). 6-min water temperature data can be accessed through [CO-OPS Data API](https://api.tidesandcurrents.noaa.gov/api/prod/).

The QA/QC’d daily water temperature data used in this notebook is available on GitHub. Each station's data is provided as a separate comma-separated values (CSV) file containing daily means, residuals, and quality control (QC) flags.

  Each CSV file includes the following columns:

  - `Date`
  - `DAILY_WATER_TEMP`
  - `residual`
  - `manual_flag`
  - `automated_flag`
  - `data_fill_flag`

  In the `manual_flag` and `automated_flag` columns, suspect data are marked with a `"1"`, while data that passed quality control remain unflagged. The `data_fill_flag` column indicates the source of any replacement data (e.g., `DCP2 E1`, `DCP3 E1`, `DCP1 E2`) used on that date. If no replacement was applied, this field is left blank.

A separate csv file, `station_metadata_list`, contains metadata for each station, including:

  - Region name and abbreviation (based on group median)
  - Data availability range
  - Data quality classification

  The **data quality classification** includes three categories:

  1. **Continuous Climatology**  
    Data are mostly continuous and suitable for climatological analysis.

  2. **Semi-Continuous**  
    Data may include significant gaps (over two years), often due to the removal of erroneous or missing records. These stations can support seasonal climatology analyses and may become suitable for more in-depth climatological work as more data are added.

  3. **Incomplete**  
    These data sets require additional, case-by-case review and are **not currently recommended** for climatological use due to various quality issues.

>   ⚠️ **Note:** Stations classified as **Incomplete** should only be used for testing and exploratory work within this notebook. They should not be used for climatological or other analysis outside this context.

## **Step 4**

- Read the csv file
- Remove flagged data
- Plot and save graphics

```python
# @title Fuctions for reading, cleaning, and plotting graphics
def clean_station_data(region, station_region_list, path):
    station_region = station_region_list[station_region_list['Group Median Regions'] == region]
    temp_dfs = {}
    resid_dfs = {}
    for i in range(len(station_region)):
        station = station_region['Station ID'].iloc[i]
        try:
            station_df = pd.read_csv(path + str(station) + '_final_daily_flagged.csv')
        except FileNotFoundError:
            print('File not found')
            continue

        try:
            station_df.loc[station_df['manual_flag'] == 1, 'DAILY_WATER_TEMP'] = np.nan
            station_df.loc[station_df['manual_flag'] == 1, 'residual'] = np.nan
        except KeyError:
            print('No manual flag for this station')

        try:
            station_df.loc[station_df['automated_flag'] == 1, 'DAILY_WATER_TEMP'] = np.nan
            station_df.loc[station_df['automated_flag'] == 1, 'residual'] = np.nan
        except KeyError:
            print('No automated flag for this station')

        temp_df = station_df[['Date', 'DAILY_WATER_TEMP']]
        resid_df = station_df[['Date', 'residual']]

        temp_df = temp_df.rename(columns={'DAILY_WATER_TEMP': station})
        resid_df = resid_df.rename(columns={'residual': station})

        temp_df['Date'] = pd.to_datetime(temp_df['Date'])
        resid_df['Date'] = pd.to_datetime(resid_df['Date'])

        temp_df = temp_df.drop_duplicates().sort_values(by='Date')
        resid_df = resid_df.drop_duplicates().sort_values(by='Date')

        temp_dfs[f"station_{i+1}"] = temp_df
        resid_dfs[f"station_{i+1}"] = resid_df

    return station_region, temp_dfs, resid_dfs

def plot_station_data(region, station_region, temp_dfs, resid_dfs):

    keys = [f"station_{i+1}" for i in range(len(station_region))]
    data_frames_temp = [temp_dfs[key] for key in keys]
    data_frames_resid = [resid_dfs[key] for key in keys]

    combined_df_temp = reduce(lambda left, right: pd.merge(left, right, on='Date', how='outer'), data_frames_temp)
    combined_df_temp = combined_df_temp.sort_values(by='Date').set_index('Date')

    combined_df_resid = reduce(lambda left, right: pd.merge(left, right, on='Date', how='outer'), data_frames_resid)
    combined_df_resid = combined_df_resid.sort_values(by='Date').set_index('Date')

    numsta = len(station_region)
    station_columns = combined_df_temp.columns[:numsta]

    traces = [go.Scatter(x=combined_df_temp.index, y=combined_df_temp[station], mode='lines', name=str(station)) for station in station_columns]
    fig = go.Figure(data=traces)
    fig.update_layout(width=1400, height=400, title=f"{station_region['Region Name'].iloc[0]} Stations Daily Temperature", title_x=0.5, yaxis_title='Temperature (°C)', xaxis_title='Years', plot_bgcolor='white')
    fig_path = f'Regional_Final_Plots/{region}'
    os.makedirs(fig_path, exist_ok=True)
    fig_file = f'{fig_path}/{region}_daily_timeseries.html'
    fig.write_html(fig_file)
    print('Figure saved: ' + fig_file)
    fig.show()

    combined_df_temp['Median'] = combined_df_temp.iloc[:, :numsta].median(axis=1)
    fig2 = px.line(combined_df_temp.reset_index(), x='Date', y='Median')
    fig2.update_layout(width=1400, height=400, title=f"{station_region['Region Name'].iloc[0]} Regional Median Temperature", title_x=0.5, yaxis_title='Temperature (°C)', xaxis_title='Years', plot_bgcolor='white')
    fig2_file = f'{fig_path}/{region}_regional_median.html'
    fig2.write_html(fig2_file)
    print('Figure saved: ' + fig2_file)
    fig2.show()

    for i in range(numsta):
        exclude_median = pd.DataFrame(combined_df_resid.iloc[:, 0])
        for j in range(numsta):
            if j != i:
                exclude_median[combined_df_resid.columns[j]] = combined_df_resid.iloc[:, j]
        combined_df_resid[f'group median w/o {combined_df_resid.columns[i]}'] = exclude_median.median(axis=1)

    for i in range(numsta):
        combined_df_resid[f'{combined_df_resid.columns[i]} - group median'] = combined_df_resid.iloc[:, i] - combined_df_resid.iloc[:, i + numsta]

    combined_df_resid['median of group medians'] = combined_df_resid.iloc[:, numsta:2 * numsta].median(axis=1)

    fig3 = px.line(combined_df_resid.reset_index(), x='Date', y=combined_df_resid.columns[numsta:2 * numsta])
    fig3.add_scatter(x=combined_df_resid.index, y=combined_df_resid['median of group medians'], mode='lines', line_color='black', name='median of group medians')
    fig3.update_layout(width=1400, height=400, title=f"{station_region['Region Name'].iloc[0]} Group Median Residuals where single station is excluded", title_x=0.5, yaxis_title='Temperature (°C)', xaxis_title='Years', plot_bgcolor='white')
    fig3_file = f'{fig_path}/{region}_regional_residual_median_plot.html'
    fig3.write_html(fig3_file)
    print('Figure saved: ' + fig3_file)
    fig3.show()

    fig4 = px.line(combined_df_resid.reset_index(), x='Date', y=combined_df_resid.columns[2 * numsta:3 * numsta])
    fig4.update_layout(width=1400, height=400, title=f"{station_region['Region Name'].iloc[0]} Station Difference from Group Median", title_x=0.5, yaxis_title='Temperature (°C)', xaxis_title='Years', plot_bgcolor='white')
    fig4_file = f'{fig_path}/{region}_regional_residual_difference_plot.html'
    fig4.write_html(fig4_file)
    print('Figure saved: ' + fig4_file)
    fig4.show()
```

```python
if region_choice == 'single':
    station_list = [region_selection]

for region in station_list:
    print(str(region))
    station_region, temp_dfs, resid_dfs = clean_station_data(region, station_region_list, path)
    plot_station_data(region, station_region, temp_dfs, resid_dfs)
```
